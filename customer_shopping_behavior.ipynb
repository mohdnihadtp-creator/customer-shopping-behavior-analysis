```python
# create a new column to age group
labels=['Young Adult','Adult','Middle-aged','Senior']
df['age_group']=pd.qcut(df['age'],q=4,labels=labels)
# qcut is used to divide the data into equal quantiles. 
# Here we are dividing the age column into 4 quantiles and labeling them as Young Adult, Adult, Middle-aged, and Senior.
```

```python
# pd.qcut(): This function divides a numeric column into quantile-based bins.
# q=4 means you want to split the age column into 4 equal-sized groups (quartiles).
# Each group will contain roughly the same number of rows, not necessarily the same age range.

# labels=labels: Instead of default numeric bin labels (0,1,2,3), you assign custom names:
# First quartile → "Young Adult"
# Second quartile → "Adult"
# Third quartile → "Middle-aged"
# Fourth quartile → "Senior"
# Result: A new column age_group is added to your DataFrame. 
# Each person’s age is categorized into one of the four groups based on which quartile their age falls into.
```

```python
# qcut: Splits data into quantiles (equal number of rows per bin).
# cut: Splits data into fixed value ranges that you define.
# Use qcut when you want balanced group sizes, and cut when you want meaningful ranges (like age brackets).
# example:-
# labels = ['Young Adult','Adult','Middle-aged','Senior']
# bins = [18, 30, 45, 60, float('inf')]   # fixed ranges
# df['age_group'] = pd.cut(df['age'], bins=bins, labels=labels, right=True)

# bins: Defines the cut points for age.
# [18, 30, 45, 60, inf] means:
# 18 ≤ age ≤ 30 → Young Adult
# 31 ≤ age ≤ 45 → Adult
# 46 ≤ age ≤ 60 → Middle-aged
# 61+ → Senior
# labels: Assigns human-readable names to each bin.
# right=True: Interval includes the right endpoint (e.g., 30 goes into Young Adult).
# You can set right=False if you want the left endpoint included instead.
```

```python
# create a column purchase frequency days.
# frequency is in text so it is best and easier to handle non text data 
# converting human-readable purchase frequency labels into standardized numeric values (days) for easier analysis and computation.
# created a Python dictionary that maps textual purchase frequency labels to the equivalent number of days.
# Example: "Weekly": 7, "Monthly": 30.
freq_mapping={'Fortnightly':14,
              'Weekly':7,
              'Monthly':30,
              'Quarterly':90,
              'Bi-Weekly':14,
              'Annually':365,
              'Every 3 Months':90}
df['purchase_frequency_days']=df['frequency_of_purchases'].map(freq_mapping)
```
